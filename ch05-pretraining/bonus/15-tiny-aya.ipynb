## 一句话

TinyAya 是面向**多语言**的小模型，架构基于 Llama 风格，重点在多语言分词和训练数据，而非架构创新。

## 特点

- **多语言分词器**：支持非拉丁语系（中文、阿拉伯语等），不像 GPT-2 那样把中文切成字节
- 小参数量（几百 M ~ 1B），适合边缘部署
- 架构与 Llama 相近（RoPE/RMSNorm/SwiGLU）

> 本 notebook 演示多语言分词对效率的影响。

```python
import tiktoken

# 对比 GPT-2 分词器对多语言的处理（低效）
tok = tiktoken.get_encoding("gpt2")
samples = {
    "英文": "Hello world, this is English.",
    "中文": "你好世界，这是中文。",
    "阿拉伯文": "مرحبا بالعالم هذا عربي",
    "日文": "こんにちは世界、これは日本語です",
}

print(f"{'语言':<8} {'字符数':<6} {'GPT-2 token 数':<14} {'每字 token'}")
print("-" * 44)
for lang, text in samples.items():
    n_chars = len(text)
    n_tokens = len(tok.encode(text))
    ratio = n_tokens / n_chars
    print(f"{lang:<8} {n_chars:<6} {n_tokens:<14} {ratio:.2f}")

print("\n💡 GPT-2 对非英文压缩率差（每字 token 多）。")
print("   TinyAya 等多语言模型用专门的分词器，让中文/阿拉伯文压缩率接近英文。")
print("   分词压缩率直接决定有效上下文长度和计算量。")
```
