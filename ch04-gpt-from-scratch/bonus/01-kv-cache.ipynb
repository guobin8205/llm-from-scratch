## 一句话

缓存历史 token 的 K/V，每步只算新 token，约 5× 加速。

## 原理

自回归生成时，前面 token 的 K/V 不会变。标准做法每步重算所有 token 的 K/V（O(seq²)）；KV 缓存存历史 K/V，每步只算新 token（O(seq)）。官方实测 124M 上约 5× 加速。

> KV 缓存是后续所有变体（GQA/MLA/KV-sharing）优化的对象。
